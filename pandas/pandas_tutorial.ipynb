```python
df2=df.copy()
df2=df2.fillna(0)
df2["age"]=df2["age"].astype("int64").copy()
df2.dtypes
```

```python
#feature engineering
df2=df.copy()
df2["age"]=df2["age"].ffill()
df2["tax"]=df["income"].apply(lambda x : "20%" if x>=60000 else "10%")
df2["Eligibility"]=df["age"].apply(lambda x : "NOT Eligible" if x>=35 else "Eligible")
gender_map={"Male":"M","Female":"F","Unknown":"UN"}
df2["gender"]=df2["gender"].map(gender_map)
df2=df2.assign(new_income=df2["income"]*1.1)
df2["country"]=df2["country"].replace("USA","US")
df2

```

```python
df2.columns=['Id', 'Name', 'Age', 'Country', 'Gender', 'Income', 'Tax',
       'Eligibility', 'New_income']
# i have updated the df here
df2=df2.rename(columns={"Income":"Salary"}) 
# df2.rename(index={1:"first"})
sorted_data=df2.sort_values("Salary").copy()
sorted_data["Ranking"]=sorted_data["Salary"].rank(method="min")
sorted_data
# sorted_data.sort_index()
# sorted_data.reset_index(drop=True)
# reorder cols
sorted_data[['Id', 'Name', 'Age', 'Country', 'Gender', 'Salary','Eligibility', 'New_income','Tax','Ranking']]
sorted_data
```

```python
df=pd.DataFrame({"country":["USA","USA","INDIA","INDIA"],
                "year":[2020,2021,2020,2021],
                "sales":[100,120,90,110],
                "profit":[20,25,18,22]})
melted_df=df.melt(
    id_vars=["country","year"],
    value_vars=["sales","profit"],
    var_name="metrics",
    value_name="value"
)
pivoted_df = melted_df.pivot(
    index=["country", "year"],
    columns="metrics",
    values="value"
)

pivoted_df
```

```python
df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Bob", "Charlie", "Dave"]
})

df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [2, 1, 4, 5],
    "amount": [250, 120, 300, 180]
})
```
