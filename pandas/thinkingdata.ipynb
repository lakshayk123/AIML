```python
import json
```

```python
# Load data
def load_data(filename):
    with open(filename,"r") as f:
        data=json.load(f)
    return data    
        
```

```python
data=load_data("store_data.json")
print(data)
print("\n")
print(type(data))
```

```python
def clean_data(data):
    text_to_num={"one":1,"two":2,"three":3,"four":4,"five":5}
    cleaned_data=[]
    unique_users=set()
    for user in data:
        #clean rating
        raw_rating=str(user["rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating=text_to_num[raw_rating]
        
        user["rating"]=raw_rating

        #handle missing value
        raw_age=user.get("age")
        if(raw_age == None):
            user['age']=None

        # deduplication handling 
        if(user["name"].strip().lower() in unique_users):
            continue
        unique_users.add(user["name"].strip().lower())  
        user["name"] = user["name"].strip()
        cleaned_data.append(user)
        
    return (cleaned_data)
```

```python
result=clean_data(data)
for user in result:
    print(user)
```

```python
#defining meaningfull insights from data
def get_insights(result):
    #avg rating
    total_rating=0
    for user in result:
        total_rating+=float(user["rating"])
    
    print(f"avg rating is= {total_rating/len(result)}")

    # poor-rating percent
    poor_rating=0
    for user in result:
        if(float(user["rating"])<3):
            poor_rating+=1
            
    print(f"poor rating is= {poor_rating/len(result) * 100}%")
```

```python
get_insights(result)
```

```python
# recommendation
def recommendation(result):
    recommendation=[]
    for user in result:
        curr_rec={}
        curr_rec["name"]=user["name"]
        if(float(user["rating"])>=4):
            curr_rec["brand"]="Apple"
        else:
            curr_rec["brand"]="samsung"

        recommendation.append(curr_rec)      
    return recommendation 
        
```

```python
print(recommendation(result))
```
